```python

# S1. Palindrome String
# Check if a string is a palindrome. Ignore case.

def is_palindrome(s):
    s = s.lower()
    if s == s[::-1]:
        print(f"Given string {s} is palindrome")
    else:
        print(f"Given string {s} is not palindrome")

s=input("Enter the string to check for palindrome")
is_palindrome(s)
```

```python
# S2. Count Vowels
# Count the number of vowels in 'Hello World'.

def vowel_cnt(s):
	count=0
	for i in s:
		if i in "aeiouAEIOU":  
			count+=1
	print("Vowel count in string is: ",count)
s=input("Enter the string to count the vowels")
vowel_cnt(s)
```

```python
# S3. Reverse Words
# Reverse each word in 'Hello World Python'.

def wrd_rev(s):
	return s[::-1]
s=input("Enter the word to reverse it: ")
print("Original word", s)
print("Word after reversing it: ",wrd_rev(s))
```

```python
# S5. Anagrams
# Check if 'listen' and 'silent' are anagrams.
# What is anagrams?
# An Anagram means two words contain exactly the same characters, with the same frequency, 
# but arranged in a different order.

def is_anagram(s1, s2):
    if len(s1) != len(s2):
        return False

    return sorted(s1) == sorted(s2)


print(is_anagram("listen", "silent"))
print(is_anagram("hello", "world"))
```

```python
# S7. Count Substring
# Count times 'the' appears in 'the cat sat on the mat and the hat'.

s='the cat sat on the mat and the hat'
s=s.split()
count=0
for i in s:
	if i=='the':
		count+=1
print("Total frequency of the in string is :",count)
```

```python
# S10. Valid Emails
# Filter valid emails (must contain @ and . after @).

emails = ["test@abc.com", "invalid.com", "hello@gmail.com",
          "user@domain", "admin@yahoo.in"]

valid_emails = []

for email in emails:
    if "@" in email and "." in email.split("@")[-1]:
        valid_emails.append(email)

print("Valid Emails:", valid_emails)
```

```python
# S11. Camel to Snake
# Convert 'camelCaseString' to 'camel_case_string'.

s = "camelCaseString"
res = ""
for i in s:
    if i.isupper():
        res += "_" + i.lower()
    else:
        res += i
print(res)
```

```python
# S12. Snake to Pascal
# Convert 'snake_case_string' to 'SnakeCaseString'.

s = 'snake_case_string'
res = ""
capitalize_next = False

for i in s:
    if i == "_":
        capitalize_next = True
    elif capitalize_next:
        res += i.upper()
        capitalize_next = False
    else:
        res += i

res = res[0].upper() + res[1:]

print(res)
```

```python
# S13. URL Params to Dict
# Parse 'order_id=101&status=success' into a dict.

url_params = 'order_id=101&status=success'

result = {}

for item in url_params.split('&'):
    key, value = item.split('=')
    result[key] = value

print(result)
```

```python
# S14. String Compression
# Compress consecutive characters:
# 'aaabbbccddddd' -> 'a3b3c2d5'.

s = 'aaabbbccddddd'

result = ''
count = 1

for i in range(1, len(s)):

    if s[i] == s[i-1]:
        count += 1

    else:
        result += s[i-1] + str(count)
        count = 1

result += s[-1] + str(count)

print(result)
```

```python
# S16. Check Digits
# Check if string contains only digits without using .isdigit().

s = input("Enter a string: ")
print("Given string is: ", s)

if s and all(i in "0123456789" for i in s):
    print("Only digits")
else:
    print("Not only digits")
```

```python
# S17. Longest Word
# Find the longest word in 'The quick brown fox jumped'.

s='The quick brown fox jumped'
s=s.split()
longest=s[0]

for i in s:
	if len(i) > len(longest):
		longest = i
print("The longest number is: ",longest)
```

```python
# S18. Most Frequent Char
# Find the most frequent character in 'banana'.

s = 'banana'

freq = {}

for i in s:
    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1

frequent = max(freq, key=freq.get)

print(frequent)
print(freq[frequent])
```

```python
# S19. Remove Duplicate Words
# Remove duplicate words preserving order.

s=" Remove Duplicate Words Remove duplicate words preserving order"
l1=s.split()
l2=[]
for i in l1:
	if i not in l2:
		l2.append(i)
s2 = " ".join(l2)
print(s2)
```
