```python
# Q28. Write a function to check whether two strings are
#      anagrams of each other.
    # -->     An anagram is a word or string formed by rearranging all the characters of another word 
    #         or string, using each character exactly the same number of times.


def anagram(a, b):
    return sorted(a) == sorted(b)

a = input("Enter first string: ").lower()
b = input("Enter second string: ").lower()

if anagram(a, b):
    print("The strings are anagrams")
else:
    print("The strings are not anagrams")
```

```python
# Q30. Write a function to remove special characters from
# a string while retaining letters, digits, and spaces.

def remove_spl(a):
    b = []
    for i in a:
        if i.isalnum() or i.isspace():
            b.append(i)
    c = "".join(b)
    return c


a=input("Enter a string")
print("The string is",a)
print("String after removing the special charcter",remove_spl(a))
```

```python
# Q31. Write a function to find common elements between
#      two lists without returning duplicates.

def com_elem(lst1,lst2):
    lst3=[]
    for i in lst1:
        if i in lst2:
            lst3.append(i)
    return lst3

lst1=list(map(int,input("Enter a list with spaces:").split()))
lst2=list(map(int,input("Enter a list with spaces:").split()))

print("The first list is", lst1)
print("The second list is",lst2)
print("Common Elemnets between list 1 and list 2",com_elem(lst1,lst2))
```

```python
# Q32. Write a function to move all zeros to the end of
#      a list while preserving the order of other elements.

def zero_move(lst1):
    lst2=[]
    for i in lst1:
        if i !=0:
            lst2.append(i)

    for i in lst1:
        if i==0:
            lst2.append(i)

    return lst2

lst1=list(map(int,input("Enter a list with spaces:").split()))
print("Original list is",lst1)
print("The lsit after moving zero to last",zero_move(lst1))
```

```python

# Q33. Write a function to find duplicate elements and
#      count how many times each occurs.

def lst_frq_cnt(lst1):
    dict = {}

    for i in lst1:
        if i not in dict:
            dict[i] = 1
        else:
            dict[i] += 1

    lst2 = []

    for j, count in dict.items():
        if count > 1:
            lst2.append((j, count))

    return lst2


lst1 = list(map(int, input("Enter a list with spaces: ").split()))

print("Original list is:", lst1)
print("Duplicate elements with count:", lst_frq_cnt(lst1))
```

```python

# Q35. Write a function to find a missing number from a list
#      containing distinct numbers from 1 to n.

def miss_num(lst7):
    b=list(range(1,int(max(lst7))))
    c=[]
    for i in b:
        if i not in lst7:
            c.append(i)
    return c


lst7 = list(map(int, input("Enter a list with between 1- 15").split()))
print("Original list",lst7)
print("Elements not present in original list between 1-15",miss_num(lst7))
    
```

```python
# Q38. Write a function to return the union, intersection,
#      and symmetric difference of two sets.

def set_ops(set1,set2):
    print("The union of two sets is:",(set1 | set2))
    print("The intersection of two sets is:", (set1 & set2))
    print("The symetric diffrence between is :",(set1 ^ set2))

set1=set(map(int,input("Enter the first set:").split()))
set2=set(map(int,input("Enter the first set:").split()))

print("The first set is:",set1)
print("The second set is:",set2)
print("The set operations are")

set_ops(set1,set2)
```

```python


# Q39. Write a function to check whether one set is a
#      subset or superset of another.

def chk_sub_supr(set3,set4):
    if set3.issubset(set4):
        print("First set is a subset of second set")
    elif set3.issuperset(set4):
        print("First set is superset of second set") 
    else:
        print("Only Sets")

set3=set(map(int,input("Enter the first set:").split()))
set4=set(map(int,input("Enter the first set:").split()))

print("The first set is:",set3)
print("The second set is:",set4)

chk_sub_supr(set3,set4)
```
