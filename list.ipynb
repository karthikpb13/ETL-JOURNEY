```python
# L1. Sort List
# Given nums = [4, 2, 7, 1, 9, 3], return the list sorted in ascending order without modifying the original.


nums = [4, 2, 7, 1, 9, 3]
print("Original list: ",nums)
print("Sorted list without modifying the original: ",sorted(nums))
print("Original list after sorting: ",nums)
```

```python

# L2. Remove Duplicates
# Remove all duplicate values from [1, 2, 2, 3, 4, 4, 5] while preserving order.

nums=[1, 2, 2, 3, 4, 4, 5]
nums2=[]
for i in nums:
	if i not in nums2:
		nums2.append(i)
print("original list: ",nums)
print("List after removing the duplicates: ",nums2)
```

```python
# L7. Count Evens
# Count how many even numbers are in [3, 6, 8, 11, 14, 17, 20].

nums=[3, 6, 8, 11, 14, 17, 20]
count=0
for i in nums:
	if i%2==0:
		count+=1
print("Total even numbers in list are: ", count)
```

```python
# L9. Nested Sum
# Find sum of elements in nested list [[1, 2], [3, [4, 5]]]

nums = [[1, 2], [3, [4, 5]]]

def nested_sum(lst):
    total = 0

    for i in lst:
        if isinstance(i, list):
            total += nested_sum(i)
        else:
            total += i

    return total


print("Sum:", nested_sum(nums))
```

```python
# L10. Max Index
# Find the index of the maximum value without index(max()).

nums=[2, 4, 6, 1, 3, 5]
max_value=nums[0]

for i in range(len(nums)):
	if nums[i] > max_value:
		max_value=nums[i]
		max_index=i
print("Maximum value present at index: ",max_index)

## here we used range(len(nums)) to get the index becasue we want index not value

```

```python
# L15. Check Palindrome List
# Check if a list [1, 2, 3, 2, 1] is a palindrome.

nums=[1, 2, 3, 2, 1]
if nums==nums[::-1]:
	print("List is palindrome")
else:
	print("List is not palindrome")
```

```python
# L16. Move Zeros
# Move all zeros to the end while keeping other elements in order.

nums=[1,0, 2,0, 3,4,5,0,6]
nums2=[]
nums3=[]
for i in nums:
	if i == 0:
		nums2.append(i)
	else:
		nums3.append(i)
print(nums3+nums2)
```

```python
# L18. Max Profit
# Find the maximum profit if you buy at one price and sell later.

prices = [7, 1, 5, 3, 6, 4]

min_price = prices[0]
max_profit = 0

for i in prices:
    if i < min_price:
        min_price = i
    else:
        profit = i - min_price

        if profit > max_profit:
            max_profit = profit

print("Maximum Profit:", max_profit)
```

```python

# L19. Batch Split
# Split [1..9] into batches of size 3.

nums = [1, 2, 3, 4, 5, 6, 7, 8, 9]

batch_size = 3

res = []

for i in range(0, len(nums), batch_size):
    res.append(nums[i:i+batch_size])

print(res)  
```

```python
# L20. Pairs
# Given ["a", "b", "c", "d"], return all possible pairs.

arr = ["a", "b", "c", "d"]

pairs = []

for i in range(len(arr)):
    for j in range(i+1, len(arr)):
        pairs.append((arr[i], arr[j]))

print(pairs)
```
